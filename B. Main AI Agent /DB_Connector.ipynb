### Databse Connector:

This logic will allow the Brain to "look" inside the database (Logistics.DB - DuckDB) to see what tables and columns are available.

```python
import duckdb

# POINT THIS TO YOUR ORIGINAL 500MB FILE
# If you have a backup of the original 500MB file, use that path here.
ORIGINAL_DB = r"B:\3. Prog\2. Projects\7. Logistics and supply chain\B. Main AI Agent\data\logistics.db"

def check_original_data():
    try:
        conn = duckdb.connect(ORIGINAL_DB, read_only=True) # Open as Read-Only to be safe
        print(f"📊 Analyzing Database: {ORIGINAL_DB}")
        
        # Get Table Names
        tables = conn.execute("SHOW TABLES").fetchall()
        print(f"Found {len(tables)} tables.")
        
        for (table_name,) in tables:
            # Get Row Count for each table
            row_count = conn.execute(f"SELECT count(*) FROM {table_name}").fetchone()[0]
            print(f"🔹 Table: {table_name} | Rows: {row_count}")
            
            # Show Columns
            cols = conn.execute(f"DESCRIBE {table_name}").fetchall()
            col_names = [c[0] for c in cols]
            print(f"   Columns: {', '.join(col_names)}\n")
            
        conn.close()
    except Exception as e:
        print(f"❌ Error reading file: {e}")

if __name__ == "__main__":
    check_original_data()
```
